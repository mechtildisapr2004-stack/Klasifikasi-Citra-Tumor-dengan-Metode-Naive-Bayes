**11. Memisahkan Data (Data Splitting)**

```python
# BAGIAN PEMISAHAN DATA (DATA SPLITTING)
X = df.drop(['Image', 'Class'], axis=1)
y = df['Class']

# Membagi data menjadi 80% Train dan 20% Test (umum digunakan dalam riset)
X_train, X_test, Y_train, Y_test = train_test_split(X, y, test_size=0.2, random_state=9)

# Menyiapkan data khusus untuk visualisasi Decision Boundary (2 fitur saja)
# Kita gunakan Homogeneity dan Skewness sesuai kode Anda sebelumnya
X_train_feature = X_train[['Homogeneity', 'Skewness']]
X_test_feature = X_test[['Homogeneity', 'Skewness']]
y_train_feature = Y_train
y_test_feature = Y_test
```

**12. Mendefinisikan Metode plot_real_pred_val**

```python
def plot_cm(y_true, y_pred, title):
    cm = confusion_matrix(y_true, y_pred)
    plt.figure(figsize=(6,5))
    sns.heatmap(cm, annot=True, fmt='d', cmap='RdPu') # Warna pink/ungu biar senada dengan gambar Anda
    plt.title(f'Confusion Matrix: {title}')
    plt.ylabel('Actual Label')
    plt.xlabel('Predicted Label')
    plt.show()

def plot_real_pred_val(y_true, y_pred, title):
    plt.figure(figsize=(12,4))
    # Mengambil 100 sampel pertama untuk perbandingan
    plt.scatter(range(len(y_true[:100])), y_true[:100], label='Real Value', alpha=0.6, marker='o')
    plt.scatter(range(len(y_pred[:100])), y_pred[:100], label='Predicted Value', alpha=0.6, marker='x')
    plt.title(f'Real vs Predicted Values - {title}')
    plt.legend()
    plt.show()
```

**13. Menerapkan Metode Naive Bayes (Menampilkan Grafik Nilai Sebenarnya vs Nilai Terprediksi dan Confusion Matrix)**

```python
# Naive Bayes model
# Defines Naive Bayes model
nbcla = GaussianNB()

# Trains model
nbcla.fit(X_train, Y_train)

# Predicts target values
pred_nb = nbcla.predict(X_test)
print(pred_nb[1:100])

# Plots real and predicted values
plot_real_pred_val(Y_test, pred_nb, "Naive Bayes")

print("\n\n")

# Plots confusion matrix
nbcla_cm = plot_cm(Y_test, pred_nb, "Naive Bayes")

print("\n\n")

# Test score
score_nbcla = nbcla.score(X_test, Y_test)
print(score_nbcla)

average_precision_score_nb = average_precision_score(Y_test, pred_nb)
print('Average precision-recall score: {0:0.2f}'.format(
    average_precision_score_nb))
```
